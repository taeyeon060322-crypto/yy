강의 08 43쪽 "수선이 끝까지 올라갈 수도 있다" 자료를 예로 들어 연쇄 수선 과정을 설명할 수 있습니다.

1차 수선 (하단 불균형 해결): 하위 트리에서 노드가 삭제되어 균형이 깨지면, 이를 해결하기 위해 첫 번째 수선(회전)을 수행합니다. 이 과정에서 해당 서브 트리의 전체 높이가 감소하게 됩니다.

2차 수선 (중단 연쇄 반응): 하위 서브 트리의 높이가 줄어들면서, 이번에는 그 상위 노드(조상 노드) 입장에서 양쪽 서브 트리의 높이 차이가 2 이상으로 벌어져 새로운 불균형이 발생합니다. 이를 해결하기 위해 두 번째 수선을 수행합니다.

3차 수선 (루트 노드 전파): 2차 수선 이후에도 서브 트리의 높이 감소가 계속 위로 전파되어, 최종적으로 최상단 루트 노드(그림의 노드 60)까지 불균형 상태가 되어 세 번째 수선(회전)이 필요해집니다.

따라서 삭제 연산 시 하위 노드에서 발생한 높이 감소가 루트 노드까지 연쇄적으로 파급될 수 있으므로, 트리의 높이가 충분히 깊다면 위로 올라가면서 3회 이상 수선해야 하는 경우가 발생할 수 있습니다.


# 14
B-트리에서 리프 노드는 모두 같은 최하단 레벨에 존재합니다. 이진 검색 트리(BST)의 성질과 마찬가지로, 내부 노드(리프가 아닌 노드)에 있는 특정 키의 '직후 원소'를 찾으려면 해당 키의 오른쪽 자식 포인터로 이동한 후, 리프 노드에 도달할 때까지 계속해서 가장 왼쪽 포인터(자식)를 따라 내려가야 합니다.
결과적으로 탐색의 종착지는 항상 가장 밑바닥 레벨인 리프 노드가 되므로, 내부 노드 원소의 직후 원소는 반드시 리프 노드에 위치하게 됩니다.
#15
노드가 가질 수 있는 최대 키의 개수를 $K$라고 하면, 자식 노드를 가리키는 페이지 번호의 개수는 $K+1$개가 됩니다.
디스크의 한 블록 크기가 8,192바이트이므로, 이 블록 안에 키와 페이지 번호가 모두 들어가기 위해 다음 부등식이 성립해야 합니다.
$$20K + 4(K+1) \le 8192$$
$$24K + 4 \le 8192$$
$$24K \le 8188$$
$$K \le 341.166...$$
따라서 B-트리의 한 노드는 최대 **341개**의 키를 가질 수 있도록 설계해야 한
#16
**정답:** 내부 노드에 있는 총 키의 수는 리프 노드의 총 개수보다 항상 1개 적다.

**풀이 과정:**
B-트리에서 내부 노드(리프가 아닌 노드)에 있는 키들은 탐색 시 하위 경로를 결정짓는 **분할자(Separator)** 역할을 합니다.
일렬로 늘어선 전체 리프 노드들의 사이사이를 구분하려면, 리프 노드의 총 개수($N_{leaf}$)보다 정확히 1개 적은 수의 분할자가 필요합니다. 따라서 전체 트리를 기준으로 내부 노드에 존재하는 키의 총합은 항상 리프 노드의 개수 - 1 이 됩니다.

#17
**정답:** 오버플로우가 생기는 경우는 없다

**풀이 과정:**
수업 자료에 따라 노드의 최대 키 개수가 $K$일 때, 일반적인 노드가 가져야 할 최소 키의 개수는 $\lfloor K/2 \rfloor$입니다. (예: K=5일 때 최소 2개)
병합(Merge)은 형제 노드에게 키를 빌려올 수 없을 때, 즉 **형제 노드 역시 최소 키 개수($\lfloor K/2 \rfloor$)만을 가지고 있을 때** 발생합니다. 언더플로우가 발생한 현재 노드의 키 개수는 $\lfloor K/2 \rfloor - 1$개입니다.

병합으로 만들어진 노드의 총 키 개수는 다음과 같이 계산됩니다.
* `(자신의 남은 키) + (형제의 키) + (부모에서 내려온 키 1개)`
* $= (\lfloor K/2 \rfloor - 1) + \lfloor K/2 \rfloor + 1$
* $= 2\lfloor K/2 \rfloor$

어떤 자연수 $K$에 대해서도 $2\lfloor K/2 \rfloor \le K$ 가 항상 성립하므로, 병합된 노드의 총 키 개수는 최대 허용치인 $K$를 절대 초과하지 않습니다. 따라서 오버플로우는 발생하지 않습니다.
#18
**정답:** 4 ~ 5

**풀이 과정:**
* 최대 키 수 $K = 1000$ $\rightarrow$ 최대 자식 수 $M = 1001$
* 최소 키 수 = $\lfloor 1000/2 \rfloor = 500$ $\rightarrow$ 최소 자식 수 $m = 501$
* 전체 키의 수 $N = 10^{12}$ (1조 개)

**1. 최소 높이 (모든 노드가 최대로 꽉 찼을 때):**
높이 $h$에서 최대로 수용 가능한 키의 수는 $(1001^h - 1)$입니다.
* $h=3$일 때: $1001^3 - 1 \approx 10^9$ (1조 개 수용 불가)
* $h=4$일 때: $1001^4 - 1 \approx 10^{12}$ 이상 (1조 개 수용 가능)
$\rightarrow$ **최소 높이 = 4**

**2. 최대 높이 (모든 노드가 최소 한도로만 찼을 때):**
루트를 제외한 모든 노드가 최소 자식 수($m=501$)만 가질 때, 높이 $h$에서 최소로 보장되는 키의 수는 $2 \times 501^{h-1} - 1$입니다.
* $h=5$일 때: $2 \times 501^4 - 1 \approx 1.25 \times 10^{11}$ (1조 개보다 작으므로, 1조 개를 담기 위해 이 높이까지 커질 수 있음)
* $h=6$일 때: $2 \times 501^5 - 1 \approx 6.3 \times 10^{13}$ (1조 개를 훨씬 초과하므로, 1조 개의 키만으로는 이 높이까지 도달 불가)
$\rightarrow$ **최대 높이 = 5**
#19
**정답:** 캐싱 기법 활용

**풀이 과정:**
B-트리의 검색은 언제나 루트 노드에서 시작하여 아래로 내려갑니다. 따라서 트리의 상층부(루트 노드와 그 직하위 레벨의 내부 노드들)는 거의 모든 탐색 경로에서 공통으로 방문하게 되므로 접근 빈도가 매우 높습니다.
이러한 **상위 레벨의 인덱스 노드들만 메인 메모리(RAM)에 캐싱(버퍼링)** 해 두면, 트리의 물리적인 구조를 전혀 바꾸지 않고도 가장 느린 작업인 디스크 I/O 횟수를 실용적이고 획기적으로 줄일 수 있습니다.
